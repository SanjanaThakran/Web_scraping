```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd 
import time
from datetime import date

WRITE_FILE_PATH = r'c:\Interest_rate_files'
CHROME_DRIVER_PATH = 'c:\\Interest_rate_files\\chromedriver-win64\\chromedriver.exe'
URL = "https://www.aubank.in/interest-rates/fixed-deposit-interest-rates" 

date1 = (date.today()).strftime('%d%m%Y') 
file_name= WRITE_FILE_PATH + '\Interest_rate_' + date1 + '.csv' 


chrome_options = Options()
chrome_options.add_argument("--headless")  # 
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")


service = Service(CHROME_DRIVER_PATH)


driver = webdriver.Chrome(executable_path=CHROME_DRIVER_PATH)

try:
    
    driver.get(URL)

    
    wait = WebDriverWait(driver, 10)
    wait.until(EC.presence_of_element_located((By.TAG_NAME, "body")))

    
    time.sleep(5)

    
    page_content = driver.page_source

    
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(page_content, 'html.parser')
    table=soup.find_all("table")
    gen_rate_table=table[0]
    sr_citizen_table=table[2]
    gen_rate_rows=gen_rate_table.find_all("tr")
    sr_citizen_rate_rows=sr_citizen_table.find_all("tr")
    
    print("Today's date =",date1)
    print("Interest rate file name = ",file_name)
    
    list_of_interest_rates = []
    tax_saver_row = []
    row_count = 0
    for i in gen_rate_rows[1:]: 
        data = i.find_all("td")
        
        row=['AU SF Bank']
        row.extend([tr.text for tr in data])
        if row_count in {0,1}: 
            row[3] = row[2]
        elif row_count == 11:
            tax_saver_row = ['AU SF Bank', 'Tax-saver Fixed Deposit', row[2], row[3]]
        list_of_interest_rates.append(row)
        if row_count == 11: list_of_interest_rates.append(tax_saver_row)
        row_count += 1
    
    row_count = 0
    for i in sr_citizen_rate_rows[1:]:  
        data = i.find_all("td")
        
        list_of_interest_rates[row_count].extend([tr.text for tr in data[1:]])
        
        if row_count in {0,1}: 
            list_of_interest_rates[row_count][5] = list_of_interest_rates[row_count][4]
        row_count += 1
   
    list_of_interest_rates[12].extend([list_of_interest_rates[11][4], list_of_interest_rates[11][5]])
  
    pd.set_option('display.max_columns', None)    
    int_rate_df=pd.DataFrame(list_of_interest_rates,columns=['bank_name','tenure','gen_rate','annualised_gen_rate',
                            'sr_rate','annualised_sr_rate'])

    int_rate_df.to_csv(file_name, mode='w', header=True, index=False)
    print("\nAU SF Bank = Success. Number of rows added = ", len(list_of_interest_rates))
finally:

    driver.quit()
```
