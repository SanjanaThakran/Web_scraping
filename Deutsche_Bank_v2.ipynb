```python
import requests #fetches data from webpage
from bs4 import BeautifulSoup #extracts data we need
import pandas as pd #to store data as csv or excel format
from datetime import date

WRITE_FILE_PATH = r'c:\Interest_rate_files'
URL = "https://www.deutschebank.co.in/en/advantage-banking/interest-rates.html"

date1 = (date.today()).strftime('%d%m%Y') # Get today's date and convert it to to dd/mm/yy format
file_name= WRITE_FILE_PATH + '\Interest_rate_' + date1 + '.csv' # Set directory path to write file

r=requests.get(URL)
#print(r)
soup=BeautifulSoup(r.text,"html.parser")
#print(soup)
table=soup.find_all("table",class_="")
table1=table[1]
rows=table1.find_all("tr")

print("Today's date =",date1)
print("Interest rate file name = ",file_name)

list_of_interest_rates = []
for i in rows[1:]:  # Skip heading
    data=i.find_all("td")
    #print(data)
    row=[tr.text for tr in data]
    list_of_interest_rates.append(row)
    
cleaned_data = [[elem.replace('\n', '').replace('\xa0', '') for elem in sublist] for sublist in list_of_interest_rates]

final_rates=[]
tax_saver_row = []
row_count = 0
for i in cleaned_data:
    x=['Deutsche Bank']
    gen_rate = i[1].strip() + '%' if i[1] else ''  # Append '%' to the rates
    sr_rate = i[2].strip() + '%' if i[2] else ''
    x.extend([i[0], gen_rate, sr_rate]) 
    x.insert(3, '')
    x.extend(['',])
    if row_count == 17: 
            tax_saver_row = ['Deutsche Bank', 'Tax-saver Fixed Deposit', x[2], x[3], x[4], x[5]]
            final_rates.append(tax_saver_row)
    final_rates.append(x)
    row_count += 1

#print(final_rates)
   
pd.set_option('display.max_columns', None)
int_rate_df=pd.DataFrame(final_rates,columns=['bank_name','tenure','gen_rate','annualised_gen_rate','sr_rate',
                                        'annualised_sr_rate'])
print(int_rate_df)
int_rate_df.to_csv(file_name, mode='a', header=False, index=False)
print("\nDeutsche Bank = Success. Number of rows added = ", len(final_rates))
```
